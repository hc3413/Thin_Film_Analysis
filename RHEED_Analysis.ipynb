# RHEED Analysis

## HC019

```python
### HC019 SSO Input ###
HC019_SSO_folder = Path(root_RHEED) / 'HC019'
HC019_SSO_file = 'HC019_SSO_STO Window Data.xls'

### RHEED Plot Settings ###
median_filter_window = 0  # Median filter window size (0 = off, odd integer to enable)
savgol_window = 15         # Savitzky-Golay window size (0 = off, odd integer to enable)
savgol_polyorder = 2      # Savitzky-Golay polynomial order (must be < savgol_window)
x_lim = (2.5,40)             # Set x-axis limits as (xmin, xmax) or None for auto
y_lim = None              # Set y-axis limits as (ymin, ymax) or None for auto

# Load the data (no header, columns: time, intensity)
HC019_SSO = pd.read_excel(HC019_SSO_folder / HC019_SSO_file, header=None, names=['Time', 'Intensity'])

# Apply median filter if enabled
if median_filter_window > 0:
    from scipy.signal import medfilt
    HC019_SSO['Intensity'] = medfilt(HC019_SSO['Intensity'], kernel_size=median_filter_window)

# Apply Savitzky-Golay filter if enabled
if savgol_window > 0:
    from scipy.signal import savgol_filter
    HC019_SSO['Intensity'] = savgol_filter(HC019_SSO['Intensity'], window_length=savgol_window, polyorder=savgol_polyorder)

# Plot
fig, ax = plt.subplots(figsize=fig_size)
ax.plot(HC019_SSO['Time'], HC019_SSO['Intensity'])
ax.set_xlabel('Time (s)')
ax.set_ylabel('Intensity (arb. units)')
ax.set_title('HC019 SSO RHEED Oscillations')
if x_lim is not None:
    ax.set_xlim(x_lim)
if y_lim is not None:
    ax.set_ylim(y_lim)
plt.tight_layout()

if export_data:
    fig.savefig(HC019_SSO_folder / f'HC019_SSO_RHEED.{fig_format}', format=fig_format, transparent=plot_transparency,dpi=600,)
plt.show()
```

# DFT

```python
dft_path = '/Users/horatiocox/Desktop/Rug_postdoc/Collaborations/Diego 2DEG/Camilo DFT/Results April 26/lvtot-re-opt.xlsx'

```

```python
dft_folder = Path(dft_path).parent
dft_file = Path(dft_path).name

### DFT Plot Settings ###
energy_offset = 0.0       # Shift energy axis: subtract this value (e.g. set to vacuum level to zero it)
x_lim = None              # Set x-axis limits as (xmin, xmax) or None for auto
y_lim = None              # Set y-axis limits as (ymin, ymax) or None for auto

# Load the data (row 0 is header: columns are index, distance(Å), energy_raw(eV/Å), energy_smooth(eV/Å))
dft_df = pd.read_excel(dft_path, header=0, usecols=[1, 2, 3, 4])
dft_df.columns = ['n', 'Distance', 'Energy_raw', 'Energy_smooth']
dft_df = dft_df.dropna(subset=['n'])  # drop header remnant
dft_df[['Distance', 'Energy_raw']] = dft_df[['Distance', 'Energy_raw']].astype(float)
dft_df['Energy_smooth'] = pd.to_numeric(dft_df['Energy_smooth'], errors='coerce')

# Apply energy offset
dft_df['Energy_raw'] = dft_df['Energy_raw'] - energy_offset
dft_df['Energy_smooth'] = dft_df['Energy_smooth'] - energy_offset

# Plot with consistent styling
fig_size_dft = set_plot_style(export_data=export_data, use_tex=True)
fig, ax = plt.subplots(figsize=fig_size_dft)

ax.plot(dft_df['Distance'], dft_df['Energy_raw'], color='C0', linewidth=0.8, alpha=0.8, label='Planar average')
ax.plot(dft_df['Distance'], dft_df['Energy_smooth'], color='C1', linewidth=1.5, label='Macroscopic average')

ax.set_xlabel(r'Distance ($\mathrm{\AA}$)')
ax.set_ylabel(r'Electrostatic potential (eV/$\mathrm{\AA}$)')

if x_lim is not None:
    ax.set_xlim(x_lim)
if y_lim is not None:
    ax.set_ylim(y_lim)

ax.legend(loc='best')
plt.tight_layout()

if export_data:
    fig.savefig(dft_folder / f'DFT_potential.{fig_format}', format=fig_format, transparent=plot_transparency, dpi=600)
plt.show()
```
