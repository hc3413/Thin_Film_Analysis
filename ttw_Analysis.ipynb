### Offset Examples for Log Scale

**How to use the `offset` parameter:**

- `offset=None` (default): No offset, plots overlay directly
- `offset=10`: Each successive plot is 1 decade (10x) higher - gives nice visual separation
- `offset=100`: Each successive plot is 2 decades (100x) higher - more separation
- `offset=5`: Each successive plot is ~0.7 decades higher - moderate separation

**Why multiplicative offsets work in log scale:**
- In log scale: log(y × 10) = log(y) + 1, so multiplying by 10 shifts up by 1 unit
- This gives uniform visual spacing regardless of the intensity values
- Both your weak film peaks and strong STO peaks remain clearly visible

### Y-Axis Labeling Convention

**Standard practice in XRD literature:**
- Y-axis: "Intensity (a.u.)" regardless of log or linear scale
- When log scale is used, the scale is evident from the axis tick formatting (10¹, 10², 10³, etc.)
- This follows standard scientific publication conventions

**When offset is applied:**
- Log scale with offset: "Log Intensity (a.u., offset)"
  - Tick marks are removed, so "Log" makes it explicit
- Linear scale with offset: "Intensity (a.u., offset)"
- In both cases, absolute values are meaningless - focus is on relative peak positions and shapes

## XRD Simulation: Complete Physical Model of Crystal Imperfections

### Critical Fix to Previous Model ❗

**The Problem with the Old Model:**
```python
# OLD (WRONG):
N_array = thickness_A / lattice_array  # N was same for all crystallites!
laue = np.sin(N_array * Q_array * c_array / 2)**2 / (np.sin(Q_array * c_array / 2)**2)
# This incorrectly averaged out the lattice parameter in the Laue function
```

**Why This Was Wrong:**
- The Laue function depends on the **phase difference**: N × Q × c
- If N and c both depend on lattice parameter 'a', but you use a constant thickness, the variations cancel out
- Result: Laue oscillations didn't respond properly to lattice parameter distributions

**The Fix:**
```python
# NEW (CORRECT):
thickness_array = [individual thickness for each crystallite]
lattice_array = [individual lattice parameter for each crystallite]
N_array = thickness_array / lattice_array  # DIFFERENT N for each!
c_array = lattice_array                     # DIFFERENT c for each!
laue = np.sin(N_array * Q_array * c_array / 2)**2 / (np.sin(Q_array * c_array / 2)**2)
```

Now each crystallite truly has its own thickness, lattice parameter, and orientation!

---

### Three Independent Sources of Disorder

#### 1. **Thickness Variation** (Surface Roughness)
- **Distribution Type**: Gaussian
- **Parameter**: `thickness_sigma_percent` (% of mean)
- **Physical Origin**: Non-uniform growth, surface roughness
- **Effect**: 
  - Changes N (number of unit cells) for each crystallite
  - Laue oscillations go out of phase → damping
  - Each crystallite has different interference pattern

#### 2. **Lattice Parameter Gradient** (Relaxation)
- **Distribution Type**: Linear gradient
- **Parameters**: `lattice_gradient = [bottom, top]` (Å)
- **Physical Origin**: Strain relaxation through film thickness, composition gradients
- **Effect**:
  - Shifts Bragg peak position (different d-spacing)
  - Changes c parameter → affects Laue phase
  - Creates asymmetric peak broadening

#### 3. **Mosaicity** (Crystallite Tilt)
- **Distribution Type**: Gaussian
- **Parameter**: `mosaicity_fwhm_deg` (degrees FWHM)
- **Physical Origin**: Grain boundaries, threading dislocations, growth defects
- **Effect**:
  - Tilted crystallites scatter at slightly different 2θ
  - Broadens peaks symmetrically
  - Does NOT affect Laue oscillations (same N and c, just rotated)

---

### Five Comparison Cases

| Case | Thickness | Lattice | Mosaicity | Color | Line |
|------|-----------|---------|-----------|-------|------|
| **Perfect** | Fixed | Fixed | 0° | Green | Solid |
| **Thickness Only** | Variable | Fixed | 0° | Blue | Dashed |
| **Lattice Only** | Fixed | Gradient | 0° | Orange | Dash-dot |
| **Mosaicity Only** | Fixed | Fixed | Variable | Purple | Dotted |
| **All Combined** | Variable | Gradient | Variable | Red | Solid |

---

### Key Physics Insights

1. **Each disorder source has a unique signature**:
   - Thickness → Laue damping, symmetric
   - Lattice gradient → Peak broadening + Laue damping
   - Mosaicity → Peak broadening only

2. **Laue oscillations now respond correctly**:
   - Thickness variation: Different N for each crystallite → phase mismatch
   - Lattice variation: Different c for each crystallite → phase mismatch
   - Both create destructive interference → fringes vanish

3. **The "All Combined" case is most realistic**:
   - Real films have all three effects
   - Shows how they compound to suppress Laue fringes
   - Matches experimental data much better

---


**Typical Values:**

- **Moderate quality**: thickness_σ ~ 5%, lattice gradient ~ 0.012 Å, mosaicity ~ 0.3°
- **Poor quality**: thickness_σ ~ 10%, lattice gradient ~ 0.05 Å, mosaicity ~ 1.0°

````python
# XRD Simulation: Complete Model with Thickness, Lattice, and Mosaicity Distributions
# Optimized for Apple Silicon M2 using NumPy's Accelerate framework

import numpy as np
import matplotlib.pyplot as plt
import time

# ========== Film Parameters ==========
lattice_BSO_mean = 4.116  # Å (BaSnO3 - mean value)
thickness_nm_mean = 20    # nm (mean thickness)
thickness_A_mean = thickness_nm_mean * 10
wavelength = 1.5405980    # Å (Cu Kα)
h, k, l = 0, 0, 2

# ========== Distribution Parameters (USER ADJUSTABLE) ==========
# Thickness variation (Gaussian) - surface roughness
roughness_nm = 1 #surface roughness in nm (distribution is normal for +/- this value from the mean)(same as for Xpert reflectivity XRR fitting)

# Lattice parameter variation (Linear gradient)
lattice_gradient = [4.07, 4.11]  # [bottom, top] - EXTREME for demonstration

# Mosaicity (Gaussian) - crystallite tilt distribution
mosaicity_fwhm_deg = 0.5  # degrees (FWHM of tilt distribution)

# Number of crystallites to simulate
n_crystallites = 10000

# Create 2theta range
d_spacing_mean = lattice_BSO_mean / l
theta_bragg_rad_mean = np.arcsin(wavelength / (2 * d_spacing_mean))
two_theta_bragg_mean = 2 * np.degrees(theta_bragg_rad_mean)
two_theta = np.linspace(two_theta_bragg_mean - 5, two_theta_bragg_mean + 5, 3000)
n_points = len(two_theta)

print(f"========== Simulation Parameters ==========")
print(f"Mean film thickness: {thickness_nm_mean} nm")
print(f"Mean lattice parameter: {lattice_BSO_mean:.4f} Å")
print(f"Bragg angle (mean): {two_theta_bragg_mean:.3f}°")
print(f"Number of crystallites: {n_crystallites}")

# ========== Define Cases ==========
cases = [
    {'name': 'Perfect Crystal', 
     'color': 'green', 'line_type': '-',
     'thickness_var': False, 'lattice_var': False, 'mosaicity_var': False},
    
    {'name': 'Thickness Only', 
     'color': 'blue', 'line_type': '--',
     'thickness_var': True, 'lattice_var': False, 'mosaicity_var': False},
    
    {'name': 'Lattice Only', 
     'color': 'orange', 'line_type': '-.',
     'thickness_var': False, 'lattice_var': True, 'mosaicity_var': False},
    
    {'name': 'Mosaicity Only', 
     'color': 'purple', 'line_type': ':',
     'thickness_var': False, 'lattice_var': False, 'mosaicity_var': True},
    
    {'name': 'All Combined', 
     'color': 'red', 'line_type': '-',
     'thickness_var': True, 'lattice_var': True, 'mosaicity_var': True}
]

# ========== Calculation Function ==========
def calculate_xrd_pattern(case):
    """
    Calculate XRD pattern with proper treatment of disorder effects.
    Each crystallite has its own thickness, lattice parameter, and orientation.
    """
    # Generate distributions for each parameter
    n = n_crystallites if any([case['thickness_var'], case['lattice_var'], case['mosaicity_var']]) else 1
    
    # 1. THICKNESS DISTRIBUTION (affects number of unit cells N)
    if case['thickness_var']:
        roughness_Angstrom= roughness_nm*10
        thickness_values = np.random.normal(thickness_A_mean, roughness_Angstrom, n)
        thickness_values = np.clip(thickness_values, thickness_A_mean * 0.5, thickness_A_mean * 1.5)
    else:
        thickness_values = np.full(n, thickness_A_mean)
    
    # 2. LATTICE PARAMETER DISTRIBUTION (affects d-spacing and N)
    if case['lattice_var']:
        # Linear gradient from bottom to top of film
        lattice_values = np.linspace(lattice_gradient[0], lattice_gradient[1], n)
    else:
        lattice_values = np.full(n, lattice_BSO_mean)
    
    # 3. MOSAICITY DISTRIBUTION (affects scattered beam direction)
    if case['mosaicity_var']:
        # Gaussian tilt distribution (FWHM to sigma: σ = FWHM / 2.355)
        mosaicity_sigma = mosaicity_fwhm_deg / 2.355
        tilt_angles_deg = np.random.normal(0, mosaicity_sigma, n)
    else:
        tilt_angles_deg = np.zeros(n)
    
    # Reshape for broadcasting: (n, 1) and (1, m) → (n, m)
    lattice_array = lattice_values[:, np.newaxis]
    thickness_array = thickness_values[:, np.newaxis]
    tilt_array = tilt_angles_deg[:, np.newaxis]
    two_theta_array = two_theta[np.newaxis, :]
    
    # Calculate d-spacing for each crystallite
    d_array = lattice_array / l
    
    # Calculate Bragg angle for each crystallite (affected by lattice parameter)
    theta_bragg_array = np.arcsin(wavelength / (2 * d_array))
    two_theta_bragg_array = 2 * np.degrees(theta_bragg_array)
    
    # Apply mosaicity: tilted crystallites have effective 2theta shift
    # A tilt changes the scattering geometry, effectively shifting the peak
    two_theta_bragg_array = two_theta_bragg_array + tilt_array
    
    # Calculate scattering vector Q for all 2theta points
    theta_array = np.radians(two_theta_array / 2)
    Q_array = (4 * np.pi / wavelength) * np.sin(theta_array)
    
    # ========== LAUE OSCILLATIONS ==========
    # Key: N and c are DIFFERENT for each crystallite!
    # N must be an integer (discrete number of unit cells)
    N_array_continuous = thickness_array / lattice_array  # Calculate N
    N_array = np.round(N_array_continuous)                # Round to nearest integer!
    c_array = lattice_array                               # Unit cell size (varies!)
    
    # Laue function: sin²(N·Q·c/2) / sin²(Q·c/2)
    # The phase (N·Q·c) is what matters - it's the total optical path difference
    epsilon = 1e-14
    phase = Q_array * c_array / 2
    laue_numerator = np.sin(N_array * phase)**2
    laue_denominator = np.sin(phase)**2 + epsilon
    laue = laue_numerator / laue_denominator
    
    # ========== BROADENING ==========
    K_scherrer = 0.9
    # Size broadening depends on actual thickness of each crystallite
    size_broadening = np.degrees((K_scherrer * wavelength) / (thickness_array * np.cos(theta_bragg_array)))
    instrumental_fwhm = 0.05
    total_fwhm = instrumental_fwhm + size_broadening
    gamma = total_fwhm / 2
    
    # Lorentzian profile (Bragg peak shape)
    lorentzian = gamma**2 / ((two_theta_array - two_theta_bragg_array)**2 + gamma**2)
    
    # Two versions:
    # 1. Bragg only (no Laue modulation)
    intensity_bragg = lorentzian.sum(axis=0)
    
    # 2. Bragg + Laue (full pattern with oscillations)
    # Each crystallite contributes with its own Laue pattern!
    intensity_laue = (laue * lorentzian).sum(axis=0)
    
    # Normalize
    intensity_bragg = intensity_bragg / intensity_bragg.max()
    intensity_laue = intensity_laue / intensity_laue.max()
    
    return {
        'name': case['name'],
        'color': case['color'],
        'line_type': case['line_type'],
        'intensity_bragg': intensity_bragg,
        'intensity_laue': intensity_laue,
        'thickness_std': thickness_values.std(),
        'lattice_std': lattice_values.std(),
        'mosaicity_std': tilt_angles_deg.std(),
        'N_mean': N_array.mean(),
        'N_std': N_array.std(),
        'N_range': (N_array.min(), N_array.max())
    }

# ========== Run Calculations ==========
print(f"\n========== Running Simulations ==========")
start_time = time.time()

results = []
for i, case in enumerate(cases):
    case_start = time.time()
    result = calculate_xrd_pattern(case)
    results.append(result)
    case_time = time.time() - case_start
    print(f"Case {i+1}/{len(cases)} - {case['name']}: {case_time:.2f}s")

elapsed_time = time.time() - start_time
print(f"\nTotal calculation time: {elapsed_time:.2f} seconds")
print(f"Average time per case: {elapsed_time/len(cases):.2f} seconds")

# ========== Plot Results ==========
fig1, ax1 = plt.subplots(1, 1, figsize=(10, 6))
fig2, ax2 = plt.subplots(1, 1, figsize=(10, 6))

# Plot experimental data first (in background)
# Extract EM007 data in the same 2theta range as simulation
# ttw_EM007.ttw_df is a list of dataframes, get the first measurement
two_theta_exp = ttw_EM007.ttw_df[0]['Angle'].values
intensity_exp = ttw_EM007.ttw_df[0]['Intensity'].values

# Find indices within plotting range
plot_range_mask = (two_theta_exp >= two_theta_bragg_mean - 5) & (two_theta_exp <= two_theta_bragg_mean + 5)
two_theta_exp_plot = two_theta_exp[plot_range_mask]
intensity_exp_plot = intensity_exp[plot_range_mask]

# Normalize experimental data
intensity_exp_normalized = intensity_exp_plot / intensity_exp_plot.max()

# Plot experimental data on both figures (in black)
ax1.plot(two_theta_exp_plot, intensity_exp_normalized, 
         linewidth=2, alpha=0.7, color='black', 
         linestyle='-', label='EM007 Experimental', zorder=1)

ax2.plot(two_theta_exp_plot, intensity_exp_normalized, 
         linewidth=2, alpha=0.7, color='black', 
         linestyle='-', label='EM007 Experimental', zorder=1)

# Plot each simulation case (on top of experimental)
for result in results:
    # Figure 1: Bragg only
    ax1.plot(two_theta, result['intensity_bragg'], 
             linewidth=2.5, alpha=0.8, color=result['color'], 
             linestyle=result['line_type'], label=result['name'], zorder=2)
    
    # Figure 2: Bragg + Laue
    ax2.plot(two_theta, result['intensity_laue'], 
             linewidth=2.5, alpha=0.8, color=result['color'], 
             linestyle=result['line_type'], label=result['name'], zorder=2)

# Figure 1: Bragg Peak Broadening
ax1.set_xlabel(r'$2\theta$ (degrees)', fontsize=12)
ax1.set_ylabel('Log Intensity (a.u.)', fontsize=12)
ax1.set_title('Bragg Peak: Effects of Crystal Imperfections', fontsize=14, fontweight='bold')
ax1.legend(loc='best', fontsize=9)
ax1.grid(True, alpha=0.3)
ax1.set_yscale('log')
ax1.set_ylim([1e-8, 2])
ax1.set_xlim([two_theta_bragg_mean - 5, two_theta_bragg_mean + 5])

# Figure 2: Laue Oscillation Damping
ax2.set_xlabel(r'$2\theta$ (degrees)', fontsize=12)
ax2.set_ylabel('Log Intensity (a.u.)', fontsize=12)
ax2.set_title('Laue Oscillations: Damping by Crystal Imperfections', fontsize=14, fontweight='bold')
ax2.legend(loc='best', fontsize=9)
ax2.grid(True, alpha=0.3)
ax2.set_yscale('log')
ax2.set_ylim([1e-14, 2])
ax2.set_xlim([two_theta_bragg_mean - 5, two_theta_bragg_mean + 5])

fig1.tight_layout()
fig2.tight_layout()
plt.show()

# ========== Summary ==========
print(f"\n========== Physical Interpretation ==========")
print(f"Crystal Imperfection Effects on XRD Patterns:\n")

for result in results:
    print(f"• {result['name']} ({result['color']}, {result['line_type']}):")
    if result['name'] == 'Perfect Crystal':
        print(f"    Sharp Bragg peak, persistent Laue fringes")
        print(f"    N = {result['N_mean']:.0f} unit cells")
    elif result['name'] == 'Thickness Only':
        print(f"    σ_thickness = {result['thickness_std']:.2f} Å")
        print(f"    N range: {result['N_range'][0]:.0f} to {result['N_range'][1]:.0f} (mean: {result['N_mean']:.0f})")
        print(f"    Laue fringes damped (different N for each crystallite)")
    elif result['name'] == 'Lattice Only':
        print(f"    σ_lattice = {result['lattice_std']:.5f} Å")
        print(f"    Lattice range: {lattice_gradient[0]:.3f} to {lattice_gradient[1]:.3f} Å")
        print(f"    N range: {result['N_range'][0]:.0f} to {result['N_range'][1]:.0f} (mean: {result['N_mean']:.0f})")
        print(f"    Bragg peak shifted + Laue damped (different N AND c)")
    elif result['name'] == 'Mosaicity Only':
        print(f"    σ_tilt = {result['mosaicity_std']:.3f}°")
        print(f"    N = {result['N_mean']:.0f} unit cells (same for all)")
        print(f"    Peaks shifted/broadened by crystallite tilt")
    else:  # All combined
        print(f"    All effects combined - realistic film")
        print(f"    σ_thickness = {result['thickness_std']:.2f} Å")
        print(f"    σ_lattice = {result['lattice_std']:.5f} Å")
        print(f"    σ_tilt = {result['mosaicity_std']:.3f}°")
        print(f"    N range: {result['N_range'][0]:.0f} to {result['N_range'][1]:.0f} (mean: {result['N_mean']:.0f})")

print(f"\n🔑 KEY INSIGHT:")
print(f"Now each crystallite properly contributes with its OWN:")
print(f"  • N (from thickness variation) → affects Laue phase")
print(f"  • c (from lattice variation) → affects Bragg position AND Laue phase")
print(f"  • orientation (mosaicity) → shifts peak position")
print(f"\n⚠️  CRITICAL: N is now rounded to nearest INTEGER")
print(f"  • N represents discrete number of coherently diffracting unit cells")
print(f"  • Lattice gradient: same thickness, different c → DIFFERENT N values!")
print(f"  • Example: 200Å film with c=4.0Å → N=50, but c=4.12Å → N=49")
print(f"  • This changes Laue fringe spacing AND phase!")
print(f"\nLaue oscillations are NOT averaged out incorrectly anymore!")
print(f"The damping comes from DESTRUCTIVE INTERFERENCE of shifted patterns.")


````

## Real-Space XRD Simulation: Layer-by-Layer Dynamical Diffraction

### Concept: Discrete Laue Function in Real Space

Instead of analytically computing the Laue function, we explicitly model the film atom-by-atom (or unit-cell-by-unit-cell) in real space.

### 1. How Roughness Works (The "Mask" Method)
You asked: *"Does the roughness change the max z value? Is the array square?"*

*   **The Array is a Box:** To use vectorized math (which is fast), we must define a fixed 3D grid ($N_x \times N_y \times N_z$). This is a "perfect box" in computer memory.
*   **The Height Map:** We generate a 2D map `height_map[x,y]` that defines the actual thickness at each spot (e.g., Column A is 48 layers, Column B is 50 layers).
*   **The Mask:** We create a boolean mask (True/False).
    *   If layer index $k < \text{height}$, `Mask = True` (Atom exists).
    *   If layer index $k \ge \text{height}$, `Mask = False` (Atom is missing).
*   **The Calculation:** We calculate waves for *every* point in the box, but then multiply by the mask.
    *   `Wave_Actual = Wave_Calculated * Mask`
    *   Missing atoms become **0**. They contribute nothing to the sum. This correctly simulates a rough surface.

### 2. How Interference is Calculated (Phase Accumulation)
You asked: *"Is the key thing the phase relative to z=0? Accumulating lattice parameter?"*

**Yes, exactly.** This is the "Cumulative Sum" method:

1.  **True Z-Position ($z_k$):**
    We don't just say $z = k \times c$. That would assume a perfect lattice.
    Instead, we sum the specific lattice parameters of all layers below:
    $$ z_k = \sum_{i=0}^{k} c_i $$
    *   If there is a strain gradient, $c_i$ changes, and $z_k$ captures the exact physical position of that unit cell.

2.  **Phase ($\phi_k$):**
    The phase of the X-ray scattered by layer $k$ depends on the total path difference ($Q \cdot z$):
    $$ \phi_k = Q \cdot z_k \cdot \cos(\text{tilt}) $$
    *   The $\cos(\text{tilt})$ term accounts for mosaicity (projection of Q).

3.  **Coherent Sum (Structure Factor $F$):**
    We sum the complex amplitudes of all layers in one column:
    $$ F_{\text{column}} = \sum_{k=0}^{N_z} \text{Mask}_k \cdot e^{i \phi_k} $$
    *   This is **Constructive/Destructive Interference**.
    *   If $c$ varies (gradient), the phases shift, and peaks broaden.
    *   If $N$ varies (roughness), the "Laue condition" ($N \cdot Q \cdot c$) breaks down, damping oscillations.

4.  **Incoherent Sum (Total Intensity):**
    We assume columns don't interfere with each other (lateral coherence length is small). We sum their intensities:
    $$ I_{\text{total}} = \sum_{\text{columns}} |F_{\text{column}}|^2 $$

### Advantages Over Analytical Approach

1.  **Physical intuition**: Real unit cells in real space
2.  **Flexibility**: Easy to add composition gradients, defects, etc.
3.  **Generality**: Can extend to RSM (off-specular) with same framework
4.  **Validation**: Can compare with analytical Laue function for perfect crystal

````python
import numpy as np
import time
from scipy.signal import convolve

# ========== 1. Core Functions: system generation, XRD calc, smoothing ==========

def smooth(y, window=5):
    """Simple moving-average smoothing (edge-preserving via 'same' conv)."""
    if window is None or window <= 1:
        return y
    kernel = np.ones(window) / window
    return convolve(y, kernel, mode='same')


def create_crystal_system(nx, ny, nz,
                          lattice_perfect,
                          lattice_gradient,
                          mode='perfect',
                          gradient_type='linear',
                          spatial_var_percent=0,
                          roughness_layers=0,
                          mosaicity_fwhm=0,
                          relaxation_length=15):
    """
    Generate lattice (per-layer c), height_map and tilt array for a set of crystallites.
    - nx,ny: lateral sampling of columns
    - nz: number of unit-cell layers in z
    - lattice_perfect: base c (Å) used for simplest 'perfect' mode
    - lattice_gradient: [c_interface, c_bulk] used by gradient/exponential profiles
    """
    # --- Lattice ---
    if mode == 'perfect':
        lattice_array = np.full((nx, ny, nz), lattice_perfect)
    elif mode == 'gradient' or mode == 'full':
        layer_indices = np.arange(nz)
        if gradient_type == 'linear':
            gradient_per_layer = (lattice_gradient[1] - lattice_gradient[0]) / float(max(nz - 1, 1))
            c_values = lattice_gradient[0] + gradient_per_layer * layer_indices
        elif gradient_type == 'exponential':
            total_change = lattice_gradient[1] - lattice_gradient[0]
            decay = 1 - np.exp(-layer_indices / float(relaxation_length))
            c_values = lattice_gradient[0] + total_change * decay
        else:
            raise ValueError(f"Unknown gradient_type: {gradient_type}")

        if mode == 'gradient':
            lattice_array = np.broadcast_to(c_values[np.newaxis, np.newaxis, :], (nx, ny, nz)).copy()
        else:
            # mode == 'full' -> allow spatial variation at z=0
            base_c = lattice_gradient[0]
            sigma = base_c * (spatial_var_percent / 100.0)
            c_start = np.random.normal(base_c, sigma, (nx, ny))
            lattice_array = c_start[:, :, np.newaxis] + c_values[np.newaxis, np.newaxis, :]
    else:
        raise ValueError(f"Unknown mode: {mode}")

    # --- Height / Roughness ---
    if mode == 'full' and roughness_layers > 0:
        heights = np.random.normal(nz, roughness_layers, (nx, ny))
        heights = np.clip(heights, 1, nz * 1.5)
        height_map = np.round(heights).astype(int)
    else:
        height_map = np.full((nx, ny), nz, dtype=int)

    # --- Mosaicity / Tilt ---
    if mode == 'full' and mosaicity_fwhm > 0:
        mosaicity_sigma = mosaicity_fwhm / 2.355
        tilt_map = np.random.normal(0.0, mosaicity_sigma, (nx, ny))
        tilt_array = np.broadcast_to(tilt_map[:, :, np.newaxis], (nx, ny, nz))
    else:
        tilt_array = np.zeros((nx, ny, nz))

    return lattice_array, height_map, tilt_array


def calculate_xrd_vectorized(lattice_array, height_map, tilt_array,
                             two_theta_range=(40,46), wavelength=1.5406,
                             attenuation_length_nm=0, instrumental_broadening=0,
                             n_points=3000, chunk_size=2000):
    """
    Compute XRD pattern from columns with optional absorption and instrumental broadening.
    Returns (two_theta, intensity_normalized).
    """
    start_time = time.time()
    nx, ny, nz = lattice_array.shape
    n_columns = nx * ny

    lattice_flat = lattice_array.reshape(n_columns, nz)
    tilt_flat = tilt_array.reshape(n_columns, nz)
    height_flat = height_map.reshape(n_columns)

    # z positions: accumulate layer thicknesses (approx c per layer -> z in Å)
    z_positions = np.cumsum(lattice_flat, axis=1)
    z_positions = np.concatenate([np.zeros((n_columns, 1)), z_positions[:, :-1]], axis=1)

    # mask of present layers per column
    idx_layer = np.arange(nz)[np.newaxis, :]
    mask = idx_layer < height_flat[:, np.newaxis]

    two_theta = np.linspace(two_theta_range[0], two_theta_range[1], n_points)
    theta = np.radians(two_theta / 2.0)
    Q_base = (4.0 * np.pi / wavelength) * np.sin(theta)

    # mosaic tilt factor (approx cos(tilt))
    tilt_rad = np.radians(tilt_flat)
    mosaicity_factor = np.cos(tilt_rad)

    # absorption: compute using a representative theta (use mean of range)
    theta_mean = np.radians((two_theta_range[0] + two_theta_range[1]) / 4.0)
    if attenuation_length_nm and attenuation_length_nm > 0:
        attenuation_A = attenuation_length_nm * 10.0  # nm -> Å
        path_length = 2.0 * z_positions / np.sin(theta_mean)
        absorption_factor = np.exp(-path_length / attenuation_A)
    else:
        absorption_factor = np.ones_like(z_positions)

    intensity_total = np.zeros(n_points, dtype=float)

    for start in range(0, n_columns, chunk_size):
        end = min(start + chunk_size, n_columns)
        z_chunk = z_positions[start:end, :]
        mask_chunk = mask[start:end, :]
        mos_chunk = mosaicity_factor[start:end, :]
        abs_chunk = absorption_factor[start:end, :]

        # phases: shape (n_points, n_columns_chunk, nz)
        phases = Q_base[:, None, None] * mos_chunk[None, :, :] * z_chunk[None, :, :]
        waves = np.exp(1j * phases) * abs_chunk[None, :, :]
        waves *= mask_chunk[None, :, :]

        F_col = np.sum(waves, axis=2)  # (n_points, n_columns_chunk)
        I_col = np.abs(F_col) ** 2
        intensity_total += np.sum(I_col, axis=1)

    # instrumental broadening
    if instrumental_broadening and instrumental_broadening > 0:
        # FIX: Calculate sigma in points correctly
        step_size = two_theta[1] - two_theta[0]
        sigma_pts = (instrumental_broadening / step_size) / 2.355
        
        # guard against extremely small sigma
        sigma_pts = max(sigma_pts, 0.5)
        
        kernel_radius = int(min(n_points // 2 - 1, max(1, int(4 * sigma_pts))))
        x_kernel = np.arange(-kernel_radius, kernel_radius + 1)
        kernel = np.exp(-x_kernel ** 2 / (2.0 * sigma_pts ** 2))
        kernel /= kernel.sum()
        intensity_total = convolve(intensity_total, kernel, mode='same')

    # normalize
    if intensity_total.max() > 0:
        intensity_total = intensity_total / intensity_total.max()

    elapsed = time.time() - start_time
    print(f"XRD calc finished in {elapsed:.3f}s (columns={n_columns})")
    return two_theta, intensity_total
````
