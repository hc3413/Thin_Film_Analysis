**How to use the `offset` parameter:**

- `offset=None` (default): No offset, plots overlay directly
- `offset=10`: Each successive plot is 1 decade (10x) higher - gives nice visual separation
- `offset=100`: Each successive plot is 2 decades (100x) higher - more separation
- `offset=5`: Each successive plot is ~0.7 decades higher - moderate separation

**Why multiplicative offsets work in log scale:**
- In log scale: log(y × 10) = log(y) + 1, so multiplying by 10 shifts up by 1 unit
- This gives uniform visual spacing regardless of the intensity values
- Both your weak film peaks and strong STO peaks remain clearly visible

- Y-axis: "Intensity (a.u.)" regardless of log or linear scale

## XRD Simulation: Complete Physical Model of Crystal Imperfections

### Critical Fix to Previous Model ❗

**The Problem with the Old Model:**
```python
# OLD (WRONG):
N_array = thickness_A / lattice_array  # N was same for all crystallites!
laue = np.sin(N_array * Q_array * c_array / 2)**2 / (np.sin(Q_array * c_array / 2)**2)
# This incorrectly averaged out the lattice parameter in the Laue function
```

**Why This Was Wrong:**
- The Laue function depends on the **phase difference**: N × Q × c
- If N and c both depend on lattice parameter 'a', but you use a constant thickness, the variations cancel out
- Result: Laue oscillations didn't respond properly to lattice parameter distributions

**The Fix:**
```python
# NEW (CORRECT):
thickness_array = [individual thickness for each crystallite]
lattice_array = [individual lattice parameter for each crystallite]
N_array = thickness_array / lattice_array  # DIFFERENT N for each!
c_array = lattice_array                     # DIFFERENT c for each!
laue = np.sin(N_array * Q_array * c_array / 2)**2 / (np.sin(Q_array * c_array / 2)**2)
```

Now each crystallite truly has its own thickness, lattice parameter, and orientation!

---

### Three Independent Sources of Disorder

#### 1. **Thickness Variation** (Surface Roughness)
- **Distribution Type**: Gaussian
- **Parameter**: `thickness_sigma_percent` (% of mean)
- **Physical Origin**: Non-uniform growth, surface roughness
- **Effect**: 
  - Changes N (number of unit cells) for each crystallite
  - Laue oscillations go out of phase → damping
  - Each crystallite has different interference pattern

#### 2. **Lattice Parameter Gradient** (Relaxation)
- **Distribution Type**: Linear gradient
- **Parameters**: `lattice_gradient = [bottom, top]` (Å)
- **Physical Origin**: Strain relaxation through film thickness, composition gradients
- **Effect**:
  - Shifts Bragg peak position (different d-spacing)
  - Changes c parameter → affects Laue phase
  - Creates asymmetric peak broadening

#### 3. **Mosaicity** (Crystallite Tilt)
- **Distribution Type**: Gaussian
- **Parameter**: `mosaicity_fwhm_deg` (degrees FWHM)
- **Physical Origin**: Grain boundaries, threading dislocations, growth defects
- **Effect**:
  - Tilted crystallites scatter at slightly different 2θ
  - Broadens peaks symmetrically
  - Does NOT affect Laue oscillations (same N and c, just rotated)

---

### Five Comparison Cases

| Case | Thickness | Lattice | Mosaicity | Color | Line |
|------|-----------|---------|-----------|-------|------|
| **Perfect** | Fixed | Fixed | 0° | Green | Solid |
| **Thickness Only** | Variable | Fixed | 0° | Blue | Dashed |
| **Lattice Only** | Fixed | Gradient | 0° | Orange | Dash-dot |
| **Mosaicity Only** | Fixed | Fixed | Variable | Purple | Dotted |
| **All Combined** | Variable | Gradient | Variable | Red | Solid |

---

### Key Physics Insights

1. **Each disorder source has a unique signature**:
   - Thickness → Laue damping, symmetric
   - Lattice gradient → Peak broadening + Laue damping
   - Mosaicity → Peak broadening only

2. **Laue oscillations now respond correctly**:
   - Thickness variation: Different N for each crystallite → phase mismatch
   - Lattice variation: Different c for each crystallite → phase mismatch
   - Both create destructive interference → fringes vanish

3. **The "All Combined" case is most realistic**:
   - Real films have all three effects
   - Shows how they compound to suppress Laue fringes
   - Matches experimental data much better

## Real-Space XRD Simulation: Layer-by-Layer Dynamical Diffraction

### Concept: Discrete Laue Function in Real Space

Instead of analytically computing the Laue function, we explicitly model the film atom-by-atom (or unit-cell-by-unit-cell) in real space.

### 1. How Roughness Works (The "Mask" Method)
You asked: *"Does the roughness change the max z value? Is the array square?"*

*   **The Array is a Box:** To use vectorized math (which is fast), we must define a fixed 3D grid ($N_x \times N_y \times N_z$). This is a "perfect box" in computer memory.
*   **The Height Map:** We generate a 2D map `height_map[x,y]` that defines the actual thickness at each spot (e.g., Column A is 48 layers, Column B is 50 layers).
*   **The Mask:** We create a boolean mask (True/False).
    *   If layer index $k < \text{height}$, `Mask = True` (Atom exists).
    *   If layer index $k \ge \text{height}$, `Mask = False` (Atom is missing).
*   **The Calculation:** We calculate waves for *every* point in the box, but then multiply by the mask.
    *   `Wave_Actual = Wave_Calculated * Mask`
    *   Missing atoms become **0**. They contribute nothing to the sum. This correctly simulates a rough surface.

### 2. How Interference is Calculated (Phase Accumulation)
You asked: *"Is the key thing the phase relative to z=0? Accumulating lattice parameter?"*

**Yes, exactly.** This is the "Cumulative Sum" method:

1.  **True Z-Position ($z_k$):**
    We don't just say $z = k \times c$. That would assume a perfect lattice.
    Instead, we sum the specific lattice parameters of all layers below:
    $$ z_k = \sum_{i=0}^{k} c_i $$
    *   If there is a strain gradient, $c_i$ changes, and $z_k$ captures the exact physical position of that unit cell.

2.  **Phase ($\phi_k$):**
    The phase of the X-ray scattered by layer $k$ depends on the total path difference ($Q \cdot z$):
    $$ \phi_k = Q \cdot z_k \cdot \cos(\text{tilt}) $$
    *   The $\cos(\text{tilt})$ term accounts for mosaicity (projection of Q).

3.  **Coherent Sum (Structure Factor $F$):**
    We sum the complex amplitudes of all layers in one column:
    $$ F_{\text{column}} = \sum_{k=0}^{N_z} \text{Mask}_k \cdot e^{i \phi_k} $$
    *   This is **Constructive/Destructive Interference**.
    *   If $c$ varies (gradient), the phases shift, and peaks broaden.
    *   If $N$ varies (roughness), the "Laue condition" ($N \cdot Q \cdot c$) breaks down, damping oscillations.

4.  **Incoherent Sum (Total Intensity):**
    We assume columns don't interfere with each other (lateral coherence length is small). We sum their intensities:
    $$ I_{\text{total}} = \sum_{\text{columns}} |F_{\text{column}}|^2 $$

### Advantages Over Analytical Approach

1.  **Physical intuition**: Real unit cells in real space
2.  **Flexibility**: Easy to add composition gradients, defects, etc.
3.  **Generality**: Can extend to RSM (off-specular) with same framework
4.  **Validation**: Can compare with analytical Laue function for perfect crystal

### 4. Spatial Variation Impact (Lateral Inhomogeneity)

Spatial variation represents lateral (x-y) fluctuations in lattice parameter across the film surface. Creates a distribution of slightly different lattice constants.

````python
# ========== Impact of Spatial Variation (Lateral Inhomogeneity) ==========

# Spatial Variation Physical Meaning:
# Each column (x,y position) starts with a SLIGHTLY DIFFERENT interface lattice parameter,
# then the SAME gradient profile is applied vertically through that column.
# 
# Example with gradient [4.10, 4.12] Å:
# - Column A: starts at 4.098 Å (interface) → relaxes to 4.118 Å (surface)
# - Column B: starts at 4.105 Å (interface) → relaxes to 4.125 Å (surface)  
# - Column C: starts at 4.102 Å (interface) → relaxes to 4.122 Å (surface)
# All columns have the SAME gradient shape, but shifted up/down.
# 
# This represents lateral composition/strain inhomogeneity across the film surface.
# 
# Implementation: A Gaussian random shift is applied to c_start for each column,
# then that shift is added to the entire vertical gradient profile (exponential/linear).
# The variation is specified as a PERCENTAGE of the base lattice parameter.
#
# NOT the same as:
# - Roughness (which varies thickness/z-height per column)
# - Mosaicity (which varies tilt angle per column)
# - Gradient (which varies c along z within each column - this is ADDED to gradient)

# Test three levels: Low, Medium, High spatial variation
spatial_levels = [
    {'value': 0.1, 'color': 'blue', 'linestyle': '-'},
    {'value': 0.25, 'color': 'orange', 'linestyle': '--'},
    {'value': 0.5, 'color': 'red', 'linestyle': '-.'}
]

print("Simulating Spatial Variation Effects...")
start_time = time.time()
spatial_results = []

for level in spatial_levels:
    t0 = time.time()
    label = f'Spatial Var {level["value"]:.2f}%'
    
    # Create crystal system with ONLY spatial variation
    lat_arr, h_map, t_arr = create_crystal_system(
        nx_viz, ny_viz, nz_viz,
        lattice_perfect=c_BSO_exp,
        lattice_gradient=[c_BSO_exp, c_BSO_exp],  # No gradient
        mode='full',  # Use full mode to enable spatial variation
        spatial_var_percent=level['value'],  # Variable spatial variation
        roughness_layers=0.0,  # No roughness
        mosaicity_fwhm=0.0  # No mosaicity
    )
    
    # Calculate XRD
    tt, intensity = calculate_xrd_vectorized(
        lat_arr, h_map, t_arr,
        two_theta_range=plot_range,
        wavelength=wavelength_viz,
        attenuation_length_nm=attenuation_viz,
        instrumental_broadening=instrumental_viz,
        n_points=2000,
        normalize=True
    )
    
    spatial_results.append({
        'name': label,
        'color': level['color'],
        'linestyle': level['linestyle'],
        'two_theta': tt,
        'intensity': intensity
    })
    
    print(f"  {label}: {time.time()-t0:.1f}s")

print(f"Total time: {time.time()-start_time:.1f}s\n")

# Plot
fig, ax = plt.subplots(figsize=(12, 7))

# Experimental data
ax.plot(x_exp[mask_exp], y_exp_norm[mask_exp], 'k-', 
        label='Experiment (EM007)', linewidth=2.5, alpha=0.6)

# Simulations
for res in spatial_results:
    ax.plot(res['two_theta'], res['intensity'], 
            color=res['color'], linestyle=res['linestyle'],
            label=res['name'], linewidth=2, alpha=0.8)

ax.set_xlabel(r'$2\theta$ (degrees)', fontsize=12)
ax.set_ylabel('Normalized Intensity', fontsize=12)
ax.set_title('Impact of Spatial Variation on XRD Pattern', fontsize=14, fontweight='bold')
ax.legend(fontsize=10)
ax.set_yscale('log')
ax.set_ylim(1e-4, 2)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("Key Observation: Higher spatial variation → Peak broadening + Laue damping")
print("Physical Reason: Different columns have different c-parameters → incoherent sum")
````

````python
# ========== 5b. Step 2: Fit Film (BSO) with Fixed Substrate Background ==========
# Purpose: Fit film parameters using STO-derived instrumental values and background
# Normalization: All data normalized to STO peak = 1. BSO peak is scaled relative to STO.

# --- 1. Generate STO Background for BSO Range ---
fit_range_BSO = (42.0, 45.5)
peak_search_range_BSO = (42.0, 45.0)
nx_fit_BSO, ny_fit_BSO = 8, 8 # Keep higher quality for film fitting

# Find BSO peak
peak_pos_BSO, peak_int_BSO = find_peak(exp_data, peak_search_range_BSO)
print(f"Found BSO peak at {peak_pos_BSO:.3f}° with intensity {peak_int_BSO:.0f}")

# Calculate Scaling Factor
# We normalize the experimental data to the STO peak (Max=1).
# The BSO peak in the experiment is much smaller.
# Ratio = BSO_Intensity / STO_Intensity
film_scale_factor = peak_int_BSO / peak_int_STO
print(f"Film/Substrate Intensity Ratio: {film_scale_factor:.4f}")

# Generate STO simulation over a WIDE range
print("\nGenerating STO background...")
# Use best fit STO parameters
nz_STO = int((p_STO[2] * 10) / p_STO[1])
lat_arr_STO, h_map_STO, t_arr_STO = create_crystal_system(
    nx_fit_STO, ny_fit_STO, nz_STO, # Use STO grid size
    p_STO[1], [p_STO[0], p_STO[1]],
    mode='full', gradient_type='exponential',
    spatial_var_percent=p_STO[3], roughness_layers=(p_STO[4]*10)/p_STO[1],
    mosaicity_fwhm=p_STO[5], relaxation_length=p_STO[6]
)

# Simulate STO over full range (BSO + STO)
full_range = (40.0, 50.0)
tt_full, int_full_STO = calculate_xrd_vectorized(
    lat_arr_STO, h_map_STO, t_arr_STO,
    two_theta_range=full_range,
    wavelength=wavelength,
    attenuation_length_nm=p_STO[7],
    instrumental_broadening=p_STO[8],
    n_points=5000,
    normalize=True # Normalize to STO peak = 1
)

# Interpolate STO signal to the BSO fitting grid (1000 points over fit_range_BSO)
tt_fit_BSO = np.linspace(fit_range_BSO[0], fit_range_BSO[1], 1000)
int_bg_STO = np.interp(tt_fit_BSO, tt_full, int_full_STO)

# Note on Attenuation: The 20nm BSO film attenuates the STO signal by <1% (negligible).
# Since we normalize to the observed STO peak, this effect is implicitly handled.

# --- 2. BSO Fitting Configuration ---
# Use STO results to constrain Instrumental Broadening and Attenuation
instr_STO = p_STO[8]
atten_STO = p_STO[7]

# Calculate c_guess for BSO
theta_rad_BSO = np.radians(peak_pos_BSO / 2)
c_guess_BSO = (wavelength / (2 * np.sin(theta_rad_BSO))) * 2

bounds_BSO = [
    (c_guess_BSO * 0.96, c_guess_BSO * 1.04), # 1. c_start: Lattice parameter at interface (Å)
    (c_guess_BSO * 0.96, c_guess_BSO * 1.04), # 2. c_end: Lattice parameter at surface (Å)
    (40, 48),   # 3. Thickness: Total film thickness (nm)
    (0.0, 0.5),     # 4. Spatial Var: Lattice variation across film (%)
    (0.0, 1.0),     # 5. Roughness: Surface roughness (nm)
    (0.0, 1.0),     # 6. Mosaicity: Crystallite tilt spread (deg)
    (1.0, 50.0),    # 7. Relaxation: Layers for strain to relax
    (atten_STO * 0.97, atten_STO * 1.03),  # 8. Attenuation: Constrained by STO fit (nm)
    (instr_STO * 0.97, instr_STO * 1.03)   # 9. Instrumental: Constrained by STO fit (deg)
]

p0_BSO = np.mean(bounds_BSO, axis=1)

# --- 3. Optimization Settings ---
# For Film: Use global search for robustness (complex Laue oscillations)
use_global_BSO = True   # Set False for faster local-only search
de_maxiter_BSO = 30     # DE iterations (reduce to 20-30 for speed)
de_popsize_BSO = 30     # DE population multiplier
powell_tol_BSO = 0.01  # Powell tolerance
powell_maxiter_BSO = 40  # Powell max iterations

# --- 4. Run BSO Optimization ---
print("\n--- Fitting BSO/SSO Film ---")
# CRITICAL: We fit the SUM of the simulated film and the fixed STO background.
# The optimizer minimizes: Residual = (Experiment - (Film_Sim * Scale + STO_Background))^2
# This correctly handles peak overlap/crossover.
res_BSO = run_xrd_fitting(exp_data, peak_int_STO, wavelength,
                          fit_range=fit_range_BSO,
                          nx_fit=nx_fit_BSO, ny_fit=ny_fit_BSO,
                          use_log_residual=True,
                          p0=p0_BSO, bounds=bounds_BSO,
                          use_global_search=use_global_BSO,
                          de_maxiter=de_maxiter_BSO,
                          de_popsize=de_popsize_BSO,
                          tol=powell_tol_BSO, 
                          maxiter=powell_maxiter_BSO,
                          background_intensity=int_bg_STO,
                          film_scale=film_scale_factor)

````

````python
# --- 4. Plot Final Result ---
p_best = res_BSO.x
print("\n" + "="*70)
print("Film (BSO/SSO) Best Fit Parameters:")
print("="*70)
print(f"  c_start (Interface):  {p_best[0]:.4f} Å (bounds: [{bounds_BSO[0][0]:.4f}, {bounds_BSO[0][1]:.4f}])")
print(f"  c_end (Surface):      {p_best[1]:.4f} Å (bounds: [{bounds_BSO[1][0]:.4f}, {bounds_BSO[1][1]:.4f}])")
print(f"  Thickness:            {p_best[2]:.2f} nm (bounds: [{bounds_BSO[2][0]:.2f}, {bounds_BSO[2][1]:.2f}])")
print(f"  Spatial Var:          {p_best[3]:.4f} % (bounds: [{bounds_BSO[3][0]:.4f}, {bounds_BSO[3][1]:.4f}])")
print(f"  Roughness:            {p_best[4]:.4f} nm (bounds: [{bounds_BSO[4][0]:.4f}, {bounds_BSO[4][1]:.4f}])")
print(f"  Mosaicity:            {p_best[5]:.4f} deg (bounds: [{bounds_BSO[5][0]:.4f}, {bounds_BSO[5][1]:.4f}])")
print(f"  Relax Length:         {p_best[6]:.2f} layers (bounds: [{bounds_BSO[6][0]:.2f}, {bounds_BSO[6][1]:.2f}])")
print(f"  Attenuation:          {p_best[7]:.0f} nm (bounds: [{bounds_BSO[7][0]:.0f}, {bounds_BSO[7][1]:.0f}])")
print(f"  Instrumental:         {p_best[8]:.4f} deg (bounds: [{bounds_BSO[8][0]:.4f}, {bounds_BSO[8][1]:.4f}])")
print("="*70)

# Re-run simulation with best parameters
nz_best = int((p_best[2] * 10) / p_best[1])
r_layers_best = (p_best[4] * 10) / p_best[1]

lat_arr, h_map, t_arr = create_crystal_system(
    nx_fit_BSO, ny_fit_BSO, nz_best,
    p_best[1], [p_best[0], p_best[1]],
    mode='full', gradient_type='exponential',
    spatial_var_percent=p_best[3], roughness_layers=r_layers_best,
    mosaicity_fwhm=p_best[5], relaxation_length=p_best[6]
)

# Calculate BSO signal
x_exp = exp_data.ttw_df[0]['Angle'].values
y_exp_norm = exp_data.ttw_df[0]['Intensity'].values / peak_int_STO # Normalize to STO=1

tt_best, int_best_BSO = calculate_xrd_vectorized(
    lat_arr, h_map, t_arr,
    two_theta_range=(x_exp.min(), x_exp.max()),
    wavelength=wavelength,
    attenuation_length_nm=p_best[7],
    instrumental_broadening=p_best[8],
    n_points=3000,
    normalize=True
)

# Add STO background (interpolated to full range)
int_bg_STO_full = np.interp(tt_best, tt_full, int_full_STO)
# Total = (BSO_Sim * Scale) + STO_Background
int_total = (int_best_BSO * film_scale_factor) + int_bg_STO_full

plt.figure(figsize=(10, 6))
plt.plot(x_exp, y_exp_norm, 'k-', label='Experiment (Norm to STO)', alpha=0.5, linewidth=2)
plt.plot(tt_best, int_total, 'r--', label='Best Fit (BSO + STO)', linewidth=2)
plt.plot(tt_best, int_bg_STO_full, 'g:', label='STO Background', linewidth=1)

plt.axvspan(fit_range_BSO[0], fit_range_BSO[1], color='yellow', alpha=0.1, label='BSO Fit Region')
plt.axvspan(fit_range_STO[0], fit_range_STO[1], color='blue', alpha=0.1, label='STO Fit Region')

plt.xlabel(r'$2\theta$ (degrees)')
plt.ylabel('Normalized Intensity (STO=1)')
plt.title('Double Fit Result: BSO Film + STO Substrate')
plt.legend()
plt.yscale('log')
#plt.ylim(1e-4, 1.5)
plt.grid(True, alpha=0.3)
plt.show()
````

- Enables proper deconvolution of overlapping contributions

### Understanding Multi-Peak Fitting

**How It Works:**

1. **Parameter Space**: Each peak has 9 parameters → Total = 18 parameters for BSO+SSO
   - BSO: [c_start, c_end, thickness, spatial_var, roughness, mosaicity, relax_len, atten, instr]
   - SSO: [c_start, c_end, thickness, spatial_var, roughness, mosaicity, relax_len, atten, instr]

2. **Objective Function**: 
   ```
   Simulated_Total = BSO_sim * scale_BSO + SSO_sim * scale_SSO + STO_background
   Residual = Σ(log(Experiment) - log(Simulated_Total))²
   ```

3. **Optimization Strategy**:
   - **Stage 1 (DE)**: Population-based search explores all 18D parameter space simultaneously
   - **Stage 2 (Powell)**: Refines the solution to high precision
   - Both peaks are **coupled** through the shared experimental data

**Key Advantages:**
- Proper deconvolution of overlapping peaks
- Shared instrumental parameters (attenuation, broadening) ensure physical consistency
- Accounts for interference between peaks in overlapping regions

**Tips for Success:**
- Set reasonable bounds based on single-peak fits
- Consider constraining attenuation/instrumental to be the same for both peaks
- Start with wider fitting ranges that capture both peaks
- Reduce `de_maxiter` or `de_popsize` for faster (but potentially less accurate) results

### Extending to More Peaks (N-Peak Fitting)

The `run_multi_peak_xrd_fitting()` function is fully generalized and can handle any number of peaks:

```python
# Example: Fitting 3 peaks (BSO, SSO, LAO)
peak_labels = ['BSO', 'SSO', 'LAO']
peak_intensities = [peak_int_BSO, peak_int_SSO, peak_int_LAO]
fit_ranges_multi = [(43.0, 45.5), (41.5, 43.5), (40.0, 42.0)]
nx_fits_multi = [8, 8, 8]
ny_fits_multi = [8, 8, 8]
bounds_list_multi = [bounds_BSO, bounds_SSO, bounds_LAO]
p0_list_multi = [p0_BSO, p0_SSO, p0_LAO]

# Run with same function call
results_multi = run_multi_peak_xrd_fitting(
    exp_data, peak_labels, peak_intensities, wavelength,
    fit_ranges_multi, nx_fits_multi, ny_fits_multi,
    p0_list=p0_list_multi, bounds_list=bounds_list_multi,
    use_global_search=True, ...
)
```

**Important Notes:**
- Each additional peak adds 9 parameters → optimization time scales significantly
- With 3 peaks (27 params), consider reducing DE iterations or using tighter bounds
- Computational cost: ~O(N²) due to increased parameter space and simulation overhead
